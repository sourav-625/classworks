```python
import numpy as np

# inputs
x = np.array([0.5, 0.2, 0.1])

# weights
w = np.array([0.4, -0.7, 0.2])

# weighted sum
y_sum = np.dot(x, w)

# simple activation (identity)
output = y_sum

print("Weighted Sum:", y_sum)
print("Output:", output)
```

```python
def biological_neuron(signals, threshold=1.0):
    soma = sum(signals) # accumulation
    if soma >= threshold:
        return 1 # neuron fires
    return 0

signals = [0.3, 0.4, 0.5]
print("Neuron Output:", biological_neuron(signals))
```

```python
import numpy as np

def artificial_neuron(inputs, weights, bias=0):
    y_sum = np.dot(inputs, weights) + bias
    return y_sum

x = np.array([1, 0, 1])
w = np.array([0.5, -0.6, 0.2])

print("Summation Output:", artificial_neuron(x, w, bias=0.1))
```

```python
import math

# Identity function
def identity(x):
    return x

# Step / Threshold function
def step(x):
    return 1 if x > 0 else 0

# ReLU function
def threshold(x, theta=0.5):
    return 1 if x >= theta else 0

# ReLU function
def relu(x):
    return max(0, x)

# Sigmoid function
def sigmoid(x, k=1):
    return 1 / (1 + math.exp(-k * x))

# Bipolar Sigmoid
def bipolar_sigmoid(x):
    return (2 / (1 + math.exp(-x))) - 1

# Hyperbolic Tangent
def tanh_n(x):
    return math.tanh(x)

# ------------- Test Values -----------------
values = [-2, -1, 0, 0.5, 1, 2]

print("Input Values:", values)
print("\n--- Identity function ---")
for v in values:
    print(v, "->", identity(v))

print("\n--- Step function ---")
for v in values:
    print(v, "->", step(v))

print("\n--- Threshold function ---")
for v in values:
    print(v, "->", threshold(v))

print("\n--- ReLU function ---")
for v in values:
    print(v, "->", relu(v))
    
print("\n--- Sigmoid function ---")
for v in values:
    print(v, "->", round(sigmoid(v), 4))

print("\n--- Bipolar Sigmoid ---")
for v in values:
    print(v, "->", round(bipolar_sigmoid(v), 4))

print("\n--- Tanh function ---")
for v in values:
    print(v, "->", round(tanh_n(v), 4))
```

```python
def mcp_neuron(x1, x2, w1=1, w2=1, threshold=1):
    y_sum = x1 * w1 + x2 * w2
    return 1 if y_sum >= threshold else 0

for x1 in [0, 1]:
    for x2 in [0, 1]:
        print(x1, x2, "->", mcp_neuron(x1, x2))
```

```python
import numpy as np

class Perceptron:
    def __init__(self, lr=0.1, epochs=10):
        self.lr = lr
        self.epochs = epochs
        
    def activation(self, x):
        return 1 if x >= 0 else -1
    
    def fit(self, X, y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0
        
        for _ in range(self.epochs):
            for i in range(len(X)):
                linear = np.dot(X[i], self.weights) + self.bias
                y_pred = self.activation(linear)
                
                update = self.lr * (y[i] - y_pred)
                self.weights += update * X[i]
                self.bias += update
                
    def predict(self, X):
        linear = np.dot(X, self.weights) + self.bias
        return self.activation(linear)

#sample data
X = np.array([[2, 3], [1, 1], [2, 1], [-1, -2]])
y = np.array([1, 1, -1, -1])

p = Perceptron()
p.fit(X, y)
print("Prediction:", p.predict(np.array([2, 2])))
```

```python
import numpy as np

class Adaline:
    def __init__(self, lr=0.01, epochs=20):
        self.lr = lr
        self.epochs = epochs
        
    def fit(self, X, y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0
        
        for _ in range(self.epochs):
            for i in range(len(X)):
                y_sum = np.dot(X[i], self.weights) + self.bias
                error = y[i] - y_sum
                
                self.weights += self.lr * error * X[i]
                self.bias += self.lr + error
                
    def predict(self, X):
        y_sum = np.dot(X, self.weights) + self.bias
        return 1 if y_sum >= 0 else -1
    
X = np.array([[1, 2], [2, 3], [3, 4], [-1, -2]])
y = np.array([1, 1, 1, -1])

a = Adaline()
a.fit(X, y)
print("Prediction:", a.predict(np.array([2, 2])))
```

```python
# Single Layer feed forward network

import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score

x,y = make_classification(n_samples = 200,n_features = 2,n_redundant = 0,n_clusters_per_class = 1)
x_tr,x_te,y_tr,y_te = train_test_split(x,y,test_size=0.3)
model= Perceptron(max_iter=1000,eta0=0.1)
model.fit(x_tr,y_tr)
y_pred = model.predict(x_te)
print("Accuracy: ",accuracy_score(y_te,y_pred))
```

```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score

x,y = make_classification(n_samples=300,n_features=4,n_redundant=0,random_state=54,n_informative=4)
x_tr,x_te,y_tr,y_te = train_test_split(x,y,test_size=0.25)
model= MLPClassifier(hidden_layer_sizes=(10,5),activation="relu",max_iter=500,learning_rate_init=0.01)
model.fit(x_tr,y_tr)
y_pred = model.predict(x_te)
print("Accuracy: ",accuracy_score(y_te,y_pred))
```

```python
transactions = [
    {"A", "B"},
    {"A", "C"},
    {"B", "c"},
    {"A", "B"},
]

min_sup = 2

tid = {}
for i, t in enumerate(transactions):
    for item in t:
        if item not in tid:
            tid[item] = set()
        tid[item].add(i)
print("TID Lists:", tid)

AB = tid["A"] & tid["B"]
print("Support of (A, B):", len(AB))

if len(AB) >= min_sup:
    print("(A, B) is frequent")
```

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.metrics import accuracy_score

X,y = make_classification(n_samples=500, n_features=5, random_state=0)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=0)
bag = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=20,
    bootstrap=True,
)
bag.fit(Xtr, ytr)
print("Bagging Accuracy:", accuracy_score(yte, bag.predict(Xte)))
```

```python
from sklearn.ensemble import AdaBoostClassifier

ada = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth = 1),
    n_estimators=50,
    learning_rate = 1.0,
    random_state = 0,
)
ada.fit(Xtr, ytr)
print("AdaBoost Accuracy:", accuracy_score(yte, ada.predict(Xte)))
```

```python
# Implementation of Gradient boosting Classifier

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score

gbm=GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
gbm.fit(Xtr,ytr)
print("GBM accuracy:",accuracy_score(yte, gbm.predict(Xte)))
```

```python
# Implementation of Elastic net
# This code shows how Elastic Net regresson

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_squared_error

# Create simple regression data
X,y=make_regression(n_samples=200, n_features=5, noise=10, random_state=0)

# Split
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.3, random_state=0)

# Elastic Net Model
model=ElasticNet(alpha=1.0, l1_ratio=0.5)
# alpha=regularization strength
# l1_ratio=balance (0=L2 only, 1=L1 only)

model.fit(Xtr, ytr)

# Predict
y_pred=model.predict(Xte)
print("MSE:",mean_squared_error(yte,y_pred))
```

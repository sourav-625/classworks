```python
#Q1

import numpy as np
import matplotlib.pyplot as plt

data = np.array([[1, 1], [2, 1], [2, 3], [4, 3], [4, 5], [6, 4], [5, 6], [6, 7], [7, 8], [8, 9]])
x = np.array([i[0] for i in data])
y = np.array([i[1] for i in data])
plt.scatter(x, y)
plt.show()
```

```python
#Q3

from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3)
labels = kmeans.fit_predict(data)
plt.scatter(data[:, 0], data[:, 1], c=labels)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], marker='x')
plt.show()
```

```python
#Q4

def mcp_neuron(x1, x2, w1=1, w2=1, threshold=1):
    y_sum = x1 * w1 + x2 * w2
    return 1 if y_sum <= threshold else 0

for x1 in [0, 1]:
    for x2 in [0, 1]:
        print(x1, x2, "->", mcp_neuron(x1, x2))
```

```python
#Q2

ar = []
for x in range(1, 6):
    kmeans = KMeans(n_clusters=x)
    labels = kmeans.fit(data)
    ar.append([x, labels.inertia_])
wcss = np.array(ar)
plt.plot(wcss[:, 0], wcss[:, 1])
plt.show()
```
